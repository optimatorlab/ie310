```python
import numpy as np
```

```python
# Write a function that uses "for" loops to multiply a mxm matrix A with a mx1 column vector x
# This would replicate how you would multiply these items "by hand"
def f(A, x):
    '''
    do stuff here
    '''
    (rows, cols) = A.shape

    b = np.empty(rows)
    for i in range(0, rows):
    	q = 0
    	for j in range(0, cols):
    		q += A[i,j] * x[j]
    	b[i] = q
	
    return b
```

```python
def createSquareMatrix(m, low=1, high=10, invertable=False):
    '''
    Returns a square matrix with integer values
    m = size of matrix
    low = smallest integer value to appear in matrix.  inclusive.
    high = largest integer value to appear in matrix.  inclusive.
    invertable = Should the generated matrix be created such that it has an inverse?
    '''
    # Generate a random mxm matrix
    random_matrix = np.random.randint(low=low, high=high+1, size=(m, m)) # low inclusive, high exclusive

    if (invertable):
        # Extract the upper triangle
        return np.triu(random_matrix)
    else:
        return random_matrix
        
    # print("\nUpper triangular matrix (top-left 5x5):\n", A[:5, :5])

def createVector(m, low=1, high=10):
    return np.random.randint(low=low, high=high+1, size=m)
```

```python
A = createSquareMatrix(90, low=1, high=10)
x = createVector(90, 1, 5)

# Show subset of elements:
A[:5, :5], x[:5]
```

```python
# Call your function and display the result
f(A,x)
```

```python
A = [[0, 7], [1, -2]]
b = [[23], [4]]
print(f'{A=}, {b=}')
```

```python
Anp = np.array(A)
bnp = np.array(b)
Anp
```

```python
def solveByInverse(A, b):
    # Use inverses to solve Ax = b
    Ainv = np.linalg.inv(A)
    x = Ainv @ b
    return x
```

```python
solveByInverse(Anp,bnp)
```

```python
def solveBySolve(A,b):
    x=np.linalg.solve(A,b)
    return x
```

```python
solveBySolve(Anp,bnp)
```

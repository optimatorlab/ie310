```python
import numpy as np
```

```python
def solveByinverse(A, b):
    Ainv = np.linalg.inv(A)
    x = Ainv@b
    return x

def solveBysolve(A, b):
    x =np.linalg.solve(A, b)
    return x
```

```python
def solveByeros(A, b):
    n = len(A)
    M = [list(A[i]) + [b[i]] for i in range(n)]
    for i in range(n):
        k = i
        while M[k][i] == 0:
            k += 1
        M[i], M[k] = M[k], M[i]
        pivot = M[i][i]
        M[i] = [m / pivot for m in M[i]]
        for k in range(n):
            if k!= i:
                factor = M[k][i]
                M[k] = [M[k][j] - factor * M[i][j] for j in range(n + 1)]
    return [row[n] for row in M]
```

```python
A = [[1, 1, 1], [0, 2, 5], [2, 5, -1]]
b = [6, -4, 27]
solveByeros(A, b)
```

```python
Anp = np.array(A)
bnp = np.array(b)
```

```python
%timeit solveByeros(A, b)
```

```python
%timeit solveByinverse(Anp, bnp)
```

```python
%timeit solveBysolve(Anp, bnp)
```
